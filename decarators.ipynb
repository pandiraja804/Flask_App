```python
import flask
```

```python
def my_deco(fun):
    def wrapper():
        print("Hi raja")
        fun()
        print("Hi pandi")
    return wrapper    
```

```python
@my_deco
def greet():
    print("hello")
```

```python
greet()
```

```python
# Login Check
```

```python
def login_Required(func):
    def wrapper():
        logged_in = True
        if logged_in:
            func()
        else:
            print("Please Login")
    return wrapper            
```

```python
@login_Required
def dashboardPage():
    print("Welcome")
```

```python
dashboardPage()
```

```python
# Function call counter
```

```python
def counter(func):
    def wrapper():
        print("Function called")
        func()
    return wrapper    
```

```python
@counter
def display():
    print("Display the data")
```

```python
display()
```
